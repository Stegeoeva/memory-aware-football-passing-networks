## 4. Expected threat from a separate training set

The 12 × 8 spatial xT model uses V = p_shot*p_goal + diag(1−p_shot)*T*V. Training excludes evaluated matches. This is retrospective match separation, not chronological forecasting. Smoothing and zero continuation after unsuccessful moves are recorded in model JSON. No VAEP/OBV or calibrated xT claim is made.

## 7. Shot-prefix xGChain / xGBuildup and literature

Each successful pre-shot participant receives the full shot xG once. Buildup excludes the shooter and key-pass provider. Credits overlap and are not divided among players. Only the prefix before each shot is used, avoiding retrospective credit to later participants.

Related methods include VAEP, xT, EPV, xGChain/xGBuildup, OBV, football flow motifs and memory networks. Exact definitions, literature and limits (`../results/extended/METHODOLOGY.md`, generated locally).
